## Generalization Error

Đề bài yêu cầu:

$$2M e^{-2\epsilon^2 N} \leq 0.03$$

$$\Rightarrow -2\epsilon^2 N \leq \ln\left(\frac{0.03}{2M}\right)$$

$$\Rightarrow N \geq \frac{-\ln\left(\frac{0.03}{2M}\right)}{2 \epsilon^2}$$


```python
import numpy as np

def calculate_the_smallest_N(epsilon, M, prob):
  return - np.log(prob / (2 *  M)) / (2 * (epsilon ** 2))
```

### Câu 1

```python
print(calculate_the_smallest_N(0.05, 1, 0.03))
```

Ta thấy đáp án gần nhất là [b] 1000

### Câu 2

```python
print(calculate_the_smallest_N(0.05, 10, 0.03))
```

Ta thấy đáp án gần nhất là [c] 1500

### Câu 3

```python
print(calculate_the_smallest_N(0.05, 100, 0.03))
```

Ta thấy đáp án gần nhất là [d] 2000

## Growth Function

Hàm tăng trưởng là hàm đa thức, có tính chất là: $1 <= m_H(N) <= 2^n$
### Câu 5
Xét i) N + 1 => Đây làm hàm tương trưởng của positive ray

Xét ii) $1 + N + \binom{N}{2}$
- Với N = 1 thì $m_H(1) = 2$, bằng với $2^1$
- Với N = 2 thì $m_H(2) = 4$, bằng với $2^2$
- Với N = 3 thì $m_H(3) = 7$, nhỏ hơn $2^3$

=> Là hàm đa thức bậc 2, thỏa tính chất hàm tăng trưởng => Đây là hàm tăng trưởng

Xét iii) $\sum_{i=1}^{|\sqrt{N}|} \binom{N}{i} $

- Với N = 1, $m_H(1) = 1$, nhỏ hơn $2^1 = 2$.
- Với N = 2, $m_H(2) = 2$, nhỏ hơn $2^2 = 4$.

=> Với N = 1, growth function chỉ có giá trị 1 là không hợp lý, vì bất kỳ tập hypothesis nào cũng phải phân chia được ít nhất 2 nhãn cho 1 điểm. Công thức này không thỏa mãn tính chất cơ bản của hàm tăng trưởng.

=> Đây không phải là hàm tăng trưởng.

Xét iv) $2^{|\sqrt{N}|}$

- Với N = 1, $m_H(1) = 2^0 = 1$, nhỏ hơn $2^1 = 2$.
- Với N = 2, $m_H(2) = 2^1 = 2$, nhỏ hơn $2^2 = 4$.
- Với N = 3, $m_H(3) = 2^1 = 2$, nhỏ hơn $2^3 = 8$.

=> Tương tự (iii), công thức này cho $m_H(1) = 1$, không hợp lý.

=> Đây không phải là hàm tăng trưởng.

Xét v) $2^N$

- Với $N = 1$, $m_H(1) = 2$, khớp.
- Với $N = 2$, $m_H(2) = 2^2 = 4$, khớp.
- Với $N = 3$, $$ m_H(3) = 8$, khớp.

=> Đây là mực chặn trên của hàm tăng trưởng nên đây là hàm tăng trưởng

**Vậy đáp án là:** [b] i, ii, v

## Fun with Intervals

### Câu 6:
Mô hình "2-intervals" chọn hai khoảng trên R, các điểm trong 2 khoảng có nhãn là 1, các điểm ngoài có nhãn là -1.

Vậy nên "2-intervals" chỉ có thể tạo tối đa 2 khoảng có nhãn +1 trên trục số.
- Với N $\leq$ 4, số khoảng có nhãn +1 tối đa có thể tạo ra là 2 nên N $\leq$ 4 không phải là breakpoint
  - Ví dụ: (1, 0, 1, 0), (0, 1, 0, 1)
- Với N = 5, ta có trường hợp (+1, -1, +1, -1, +1), tạo ra 3 khoảng có nhãn +1.

Vậy nên k = 5 là breakpoint của "2-intervals"

**Kết quả là:** [c] 5

### Câu 7:
Với "2-intervals", ta có các trường hợp tại ra số khoảng có nhãn +1 là
- 0 đoạn: Có 1 cách
- 1 đoạn: Tương tự như "1-intervals" (hay còn gọi như trong video bài giảng là positive intervals). Có $\binom{N + 1}{2}$ cách
- 2 đoạn: Chọn 2 đoạn không giao nhau. Tương đương với chọn 4 điểm mút (2 điểm mút cho mỗi đoạn). Có $\binom{N + 1}{4}$ cách

Vậy nên hàm tăng trưởng có dạng
- Với N > 4 thì hàm tăng trưởng có dạng:
$$
m_H(N) = 1 + \binom{N + 1}{2} + \binom{N + 1}{4}   (N >4)
$$

- Với N <= 4 thì hàm tăng trưởng có dạng:
$$
m_H(N) = 2^N  (vì breakpoint của 2-intervals là k = 5)
$$

**Vậy đáp án là:** [c] $1 + \binom{N + 1}{2} + \binom{N + 1}{4}$

## Non-Convex Sets: Concentric Circles

### Câu 10

Với N điểm $(x_1, x_2, ..., x_n)$, mỗi điểm $x_i = (x_{1i}, x_{2i})$ nên ta có khoảng cách từ gốc tọa độ tới điểm $x_i$ là
$$
r_i = \sqrt{x_{1i}^2 + x_{2i}^2}
$$

$$
a \leq r_i \leq b \quad (a, b, r_i \geq  0)
$$

Vậy bài toán quay về dạng "1-intervals" với giả sử là không có điểm nó có cùng khoảng cách tới gốc tọa độ (tức $r_i$ khác nhau từng đôi một). Khi đó các điểm trong [a, b] sẽ có nhãn là +1, ngược lại có nhãn là -1

Như ta đã biết, hàm tăng trưởng của "1-intervals" (công thức đã chứng minh trong video bài giảng với tên gọi là positive intervals) là:
$$
\binom{N + 1}{2} + 1
$$
Trong đó:
**Như vậy đáp án là** [b] $\binom{N + 1}{2} + 1$
