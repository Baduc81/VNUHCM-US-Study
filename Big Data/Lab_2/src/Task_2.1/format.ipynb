Có thể thấy là file được chia ra làm 2 phần riêng biệt, bên trái là 3 cột đáp án, bên phải là giá trị null.

Do đó em sẽ thực hiện các bước xử lý như sau để định dạng lại kết quả:
- 1: Loại bỏ các giá trị `null` ở cuối dòng.
- 2: Tách các giá trị ở bên trái ra thành các cột riêng biệt.
- 3: Thêm vào từng dòng vào `dictionary` và sau đó lưu xuống csv bằng pandas.

```python
d = {
    "report_date": [],
    "category": [],
    "revenue": [],
}

with open('out.txt', 'r') as f:
    lines = f.read().split('\n')
    for line in lines:
        line = line.strip()
        if line == '':
            continue

        # seperate the 3 columns and null        
        key, _ = line.split('\t')
        
        # remove '[' and ']'
        key = key[1: -1]

        report_date, category, revenue = list(csv.reader([key]))[0]
        d['report_date'].append(report_date.strip().replace('"', ''))
        d['category'].append(category.strip().replace('"', ''))
        d['revenue'].append(revenue.strip().replace('"', ''))

df = pd.DataFrame(d)
```

Có thể thấy dữ liệu đã đúng format tuy nhiên vẫn chưa được sort. Để sort được theo `report_date` thì em sẽ convert nó qua kiêu `datetime`.

```python
df['report_date'] = pd.to_datetime(df['report_date'], format='%d/%m/%Y')
df
```

```python
df = df.sort_values(by=['report_date', 'category'])
df
```

Bây giờ dữ liệu đã được sort, tuy nhiên cột `report_date` lại đang bị sai format D:, do đó em sẽ convert lại từ `datetime` sang `str`.

```python
df['report_date'] = df['report_date'].dt.strftime('%d/%m/%Y')
df
```

Như vậy việc xử lý dữ liệu cuối cùng cũng xong. Vậy việc cuối cùng cần làm chính là lưu xuống file csv. 

```python
df.set_index('report_date').to_csv('output.csv')
```
